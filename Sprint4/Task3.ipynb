```
public interface ICustomer
{
    int CustomerId { get; }
    string Name { get; }
    string Email { get; set; }
    void ViewProfile();
    string GetFullName();
    void UpdateEmail(string newEmail);
}

public abstract class Customer : ICustomer
{
    private string _email;

    public int CustomerId { get; protected set; }
    public string Name { get; protected set; }

    public string Email
    {
        get => _email;
        set
        {
            if (string.IsNullOrWhiteSpace(value) || !value.Contains("@"))
                throw new ArgumentException("Некорректный email.");
            _email = value;
        }
    }

    public Customer(int customerId, string name, string email)
    {
        if (customerId <= 0) throw new ArgumentException("ID клиента должен быть положительным.");
        if (string.IsNullOrWhiteSpace(name)) throw new ArgumentException("Имя не может быть пустым.");
        CustomerId = customerId;
        Name = name;
        Email = email;
    }

    public virtual string GetFullName() => Name;

    public virtual void UpdateEmail(string newEmail)
    {
        Email = newEmail;
        Console.WriteLine($"📧 Email обновлен на: {Email}");
    }

    public virtual void ViewProfile()
    {
        Console.WriteLine($"🆔 ID: {CustomerId}");
        Console.WriteLine($"👤 Имя: {GetFullName()}");
        Console.WriteLine($"✉️ Email: {Email}");
    }
}

public class VipCustomer : Customer
{
    public int LoyaltyPoints { get; private set; }
    public string VipTier { get; private set; }
    public DateTime VipSince { get; private set; }
    public bool HasPersonalManager { get; private set; }

    public VipCustomer(int customerId, string name, string email, int loyaltyPoints)
        : base(customerId, name, email)
    {
        if (loyaltyPoints < 0) throw new ArgumentException("Баллы лояльности не могут быть отрицательными.");
        LoyaltyPoints = loyaltyPoints;
        VipTier = loyaltyPoints > 1000 ? "Platinum" : "Gold";
        VipSince = DateTime.Now.AddYears(-1);
        HasPersonalManager = loyaltyPoints > 500;
    }

    public void AddLoyaltyPoints(int points)
    {
        LoyaltyPoints += points;
        Console.WriteLine($"🎁 +{points} баллов. Всего: {LoyaltyPoints}");
    }

    public void RequestConcierge()
    {
        Console.WriteLine($"{Name} вызывает персонального консьержа!");
    }

    public override string GetFullName()
    {
        return $"{Name} (VIP)";
    }

    public override void ViewProfile()
    {
        base.ViewProfile();
        Console.WriteLine($"💎 Баланс лояльности: {LoyaltyPoints}");
        Console.WriteLine($"🏅 Уровень VIP: {VipTier}");
        Console.WriteLine($"📆 VIP с: {VipSince:yyyy-MM-dd}");
        Console.WriteLine($"👨‍💼 Персональный менеджер: {(HasPersonalManager ? "Да" : "Нет")}");
    }
}

public class RegularCustomer : Customer
{
    public DateTime RegistrationDate { get; protected set; }
    public DateTime? LastEmailUpdate { get; private set; }
    public int TotalPurchases { get; set; }
    public string PreferredCategory { get; set; } = "General";

    public RegularCustomer(int customerId, string name, string email, DateTime registrationDate)
        : base(customerId, name, email)
    {
        RegistrationDate = registrationDate;
    }

    public void MakePurchase(decimal amount)
    {
        TotalPurchases++;
        Console.WriteLine($"{Name} совершил покупку на {amount:C}. Всего покупок: {TotalPurchases}");
    }

    public override void UpdateEmail(string newEmail)
    {
        base.UpdateEmail(newEmail);
        LastEmailUpdate = DateTime.Now;
        Console.WriteLine($"📅 Дата последнего обновления email: {LastEmailUpdate:yyyy-MM-dd HH:mm}");
    }

    public override void ViewProfile()
    {
        base.ViewProfile();
        Console.WriteLine($"🗓️ Дата регистрации: {RegistrationDate:yyyy-MM-dd}");
        Console.WriteLine($"🛒 Всего покупок: {TotalPurchases}");
        Console.WriteLine($"📂 Предпочтительная категория: {PreferredCategory}");
        if (LastEmailUpdate.HasValue)
            Console.WriteLine($"📬 Последнее обновление email: {LastEmailUpdate:yyyy-MM-dd HH:mm}");
    }
}

public class GroupCustomer : Customer
{
    public string GroupName { get; protected set; }
    public int MaxMembers { get; private set; } = 50;
    public DateTime CreationDate { get; private set; }
    private List<Customer> _members = new();

    public IReadOnlyList<Customer> Members => _members.AsReadOnly();

    public GroupCustomer(int customerId, string groupName, string email)
        : base(customerId, groupName, email)
    {
        GroupName = groupName;
        CreationDate = DateTime.Now;
    }

    public override string GetFullName()
    {
        return $"[ГРУППА] {GroupName} (участников: {_members.Count})";
    }

    public void AddMember(Customer member)
    {
        if (member == null) throw new ArgumentNullException(nameof(member));
        if (_members.Count >= MaxMembers)
        {
            Console.WriteLine($"❌ Группа '{GroupName}' достигла лимита участников ({MaxMembers}).");
            return;
        }
        if (_members.Contains(member))
        {
            Console.WriteLine($"⚠️ {member.Name} уже в группе.");
            return;
        }
        _members.Add(member);
        Console.WriteLine($"✅ {member.Name} добавлен в группу '{GroupName}'");
    }

    public void RemoveMember(Customer member)
    {
        if (_members.Remove(member))
        {
            Console.WriteLine($"🗑️ {member.Name} удалён из группы '{GroupName}'");
        }
    }

    public override void ViewProfile()
    {
        Console.WriteLine($"🆔 ID группы: {CustomerId}");
        Console.WriteLine($"🏷️ Название: Группа '{GroupName}'");
        Console.WriteLine($"✉️ Контактный email: {Email}");
        Console.WriteLine($"👥 Участников: {_members.Count} / {MaxMembers}");
        Console.WriteLine($"📆 Создана: {CreationDate:yyyy-MM-dd}");

        if (_members.Any())
        {
            Console.WriteLine("📋 Список участников:");
            foreach (var m in _members)
                Console.WriteLine($"  → {m.Name} (ID: {m.CustomerId})");
        }
    }
}

public class CustomerService
{
    private readonly ICustomer _customer;

    public CustomerService(ICustomer customer)
    {
        _customer = customer;
    }

    public void DisplayInfo()
    {
        Console.WriteLine("🔍 Информация от сервиса:");
        _customer.ViewProfile();
        Console.WriteLine($"📝 Полное имя: {_customer.GetFullName()}");
        Console.WriteLine();
    }

    public void SafeEmailUpdate(string newEmail)
    {
        try
        {
            _customer.UpdateEmail(newEmail);
        }
        catch (ArgumentException ex)
        {
            Console.WriteLine($"❌ Не удалось обновить email: {ex.Message}");
        }
    }
}

var vip = new VipCustomer(1, "Алексей", "alex@vip.com", 800);
var regular = new RegularCustomer(2, "Мария", "maria@test.com", new DateTime(2024, 1, 15));
var group = new GroupCustomer(100, "Команда Alpha", "alpha@org.com");

group.AddMember(vip);
group.AddMember(regular);

ICustomer[] customers = { vip, regular, group };

foreach (var cust in customers)
{
    var service = new CustomerService(cust);
    service.DisplayInfo();
}
```
